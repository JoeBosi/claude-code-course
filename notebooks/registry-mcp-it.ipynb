# Interroga il registry ufficiale dei server MCP
Modulo *Ecosistema*. Il registry ufficiale ha un'API pubblica, senza chiave e senza account: la chiami con `curl` o con Python. In questo notebook lo fai in due modi.

Esegui le celle dall'alto verso il basso (Maiusc+Invio su ognuna).

## 1. curl + jq

```python
%%bash
which jq >/dev/null || (apt-get -qq update && apt-get -qq install -y jq >/dev/null)
curl -s "https://registry.modelcontextprotocol.io/v0/servers?limit=100&search=postgres" \
  | jq -r '.servers[].server.name' | sort -u
```

Un elenco di nomi nel formato `dominio/nome-del-server`. L'elenco cambia nel tempo: conta il formato, non i nomi esatti.

## 2. Python

Molte API rifiutano le richieste senza un `User-Agent` (la «firma» del programma che chiama) e rispondono 403. Questa no, ma indicarlo è una buona abitudine: costa una riga.

```python
%pip install -q requests
import requests

URL = "https://registry.modelcontextprotocol.io/v0/servers"
risposta = requests.get(
    URL,
    params={"limit": 100, "search": "dbhub"},
    headers={"User-Agent": "claude-code-course-notebook/1.0"},
    timeout=30,
)
print("HTTP", risposta.status_code)
nomi = sorted({s["server"]["name"] for s in risposta.json()["servers"]})
for nome in nomi:
    print(nome)
```

Vedi `HTTP 200` e, tra i nomi, `io.github.bytebase/dbhub`: è il server DBHub scelto nell'esempio del modulo.
