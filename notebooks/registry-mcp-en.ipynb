# Query the official MCP server registry
Module *Ecosystem*. The official registry has a public API, no key and no account: you call it with `curl` or with Python. In this notebook you do it in two ways.

Run the cells from top to bottom (Shift+Enter on each).

## 1. curl + jq

```python
%%bash
which jq >/dev/null || (apt-get -qq update && apt-get -qq install -y jq >/dev/null)
curl -s "https://registry.modelcontextprotocol.io/v0/servers?limit=100&search=postgres" \
  | jq -r '.servers[].server.name' | sort -u
```

A list of names in the format `domain/server-name`. The list changes over time: what matters is the format, not the exact names.

## 2. Python

Many APIs reject requests without a `User-Agent` (the "signature" of the calling program) and answer 403. This one does not, but setting it is a good habit: it costs one line.

```python
%pip install -q requests
import requests

URL = "https://registry.modelcontextprotocol.io/v0/servers"
response = requests.get(
    URL,
    params={"limit": 100, "search": "dbhub"},
    headers={"User-Agent": "claude-code-course-notebook/1.0"},
    timeout=30,
)
print("HTTP", response.status_code)
names = sorted({s["server"]["name"] for s in response.json()["servers"]})
for name in names:
    print(name)
```

You see `HTTP 200` and, among the names, `io.github.bytebase/dbhub`: the DBHub server chosen in the module's example.
