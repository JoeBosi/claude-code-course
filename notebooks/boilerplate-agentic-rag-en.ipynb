# Download the agentic-rag boilerplate and run its tests
Module *Boilerplate*, guided example (steps 1-4). You download the zip from the course site, initialise it and run the seven Python tests. No Claude Code and no account needed.

Run the cells from top to bottom (Shift+Enter on each).

```python
%%bash
rm -rf /tmp/boiler && mkdir /tmp/boiler && cd /tmp/boiler
curl -fsSL -O https://claude-tutorial.giuseppebosi.com/downloads/agentic-rag.zip
unzip -q agentic-rag.zip && cd agentic-rag
ls -A
```

```python
%%bash
cd /tmp/boiler/agentic-rag
./init.sh "internal-kb" "Internal knowledge base on runbooks"
```

```python
%pip install -q uv
```

```python
%%bash
cd /tmp/boiler/agentic-rag
uv sync 2>&1 | tail -3
uv run pytest
```

At the bottom you see `7 passed`: seven tests, including the one that checks that the domain does not import external libraries.
